##  Prueba con partículas

Crea `prueba_particula.py`:

```python
from particle import Particle

electron = Particle.from_pdgid(11)
muon = Particle.from_pdgid(13)
top = Particle.from_pdgid(6)

print(electron)
print(muon)
print(top)
```

`Particle.from_pdgid(...)` identifica una partícula mediante su **PDG ID**:

```text
11 → electrón
13 → muón
 6 → quark top
```

Ejecuta:

```bash
python prueba_particula.py
```

## Prueba con cuadrivectores

Crea `prueba_vector.py`:

```python
import vector

p = vector.obj(
    px=30.0,
    py=40.0,
    pz=100.0,
    E=120.0,
)

print("pT =", p.pt)
print("|p| =", p.p)
print("masa invariante =", p.mass)
```

`vector.obj(...)` construye un cuadrivector a partir de las componentes del momento y la energía.

```text
p.pt   → momento transversal
p.p    → magnitud del trimomento
p.mass → masa invariante
```

Ejecuta:

```bash
python prueba_vector.py
```

muestra los paquetes registrados en el ambiente Conda.

```bash
python -m pip list
```

muestra los paquetes reconocidos por `pip` para el Python activo.

---

##  Guardar el ambiente

```bash
conda env export > environment.yml
```

exporta la configuración del ambiente al archivo `environment.yml`.

El símbolo:

```text
>
```

redirige la salida hacia un archivo.

Para una versión más portable:

```bash
conda env export --from-history > environment-minimal.yml
```

`--from-history` conserva principalmente los paquetes solicitados explícitamente.

---

## 24. `environment.yml` recomendado

```yaml
name: fisica-computacional

channels:
  - conda-forge
  - defaults

dependencies:
  - python=3.10
  - numpy
  - scipy
  - matplotlib
  - pandas
  - sympy
  - scikit-learn
  - jupyterlab
  - notebook
  - ipykernel
  - pip
  - pip:
      - uproot
      - awkward
      - vector
      - hist
      - mplhep
      - particle
      - iminuit
      - uncertainties
      - tqdm
      - rich
      - pytest
      - ruff
      - black
```

Las partes principales son:

```text
name         → nombre del ambiente
channels     → fuentes de paquetes
dependencies → Python y bibliotecas requeridas
pip          → paquetes instalados mediante pip
```

El archivo funciona como una **descripción reproducible del entorno**.

---

##  Crear el ambiente desde `environment.yml`

Ejecuta:

```bash
conda env create -f environment.yml
```

La opción:

```text
-f environment.yml
```

indica qué archivo contiene la definición del ambiente.

Después:

```bash
conda activate fisica-computacional
```

---

##  Actualizar el ambiente



```powershell
conda env update --name fisica-computacional --file environment.yml --prune
```

Las opciones indican:

```text
--name → ambiente que se actualizará
--file → archivo de configuración
--prune → elimina dependencias que ya no están declaradas
```
